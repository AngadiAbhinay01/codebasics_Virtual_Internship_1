##  4. Interquartile Range (IQR) of Customer Spending

###  Objective

Calculate the **Interquartile Range (IQR)** of `amount_spend` to understand the spread of the **middle 50% of customer spending values**.

###  Formula

$$
IQR = Q3 - Q1
$$

Where:

- **Q1 (25th percentile):** 25% of customers spend less than or equal to this amount.
- **Q3 (75th percentile):** 75% of customers spend less than or equal to this amount.
- **IQR:** Represents the range covered by the middle 50% of customer spending.

###  Business Objective

The IQR helps us understand the **typical spread of customer spending** while being less affected by extreme spending values or outliers.

```python
# Calculate Q1 and Q3
Q1 = df['amount_spend'].quantile(0.25)
Q3 = df['amount_spend'].quantile(0.75)
```

```python
# Calculate IQR
IQR = Q3 - Q1

print("Q1 (25th percentile):", Q1)
print("Q3 (75th percentile):", Q3)
print("IQR:", IQR)
```

$$

$$

##  6. Z-Score of a Customer Spending $450

###  Objective

Determine how an individual customer's spending compares to the **average customer spending** by calculating the **z-score** for a customer who spends **$450**.

A z-score tells us how many **standard deviations** a particular value is away from the mean.

###  Z-Score Formula

$$
z = \frac{x - \mu}{\sigma}
$$

Where:

- **$x$** = Customer's spending = **$450**
- **$\mu$** = Mean customer spending
- **$\sigma$** = Standard deviation of customer spending

###  Interpretation

- **z = 0** → Customer spends exactly at the average level.
- **Positive z-score** → Customer spends more than the average.
- **Negative z-score** → Customer spends less than the average.
- **Higher positive z-score** → Customer is significantly above the average spending level.

A customer with a **high positive z-score** can be considered a potential **high-value customer**, as their spending is significantly higher than the typical customer.

```python
# Calculate mean and standard deviation
mean_spend = df['amount_spend'].mean()
std_spend = df['amount_spend'].std()
```

```python
# Customer spending
customer_spend = 450
```

```python
# Calculate z-score
z_score = (customer_spend - mean_spend) / std_spend
```

```python
print("Mean spending:", mean_spend)
print("Standard deviation:", std_spend)
print("Customer spending:", customer_spend)
print("Z-score:", z_score)
```

## 📊 Interpretation

The mean customer spending is $238.23, and the standard deviation is $155.94.

For a customer who spends **$450**, the calculated z-score is:

$$
z = \frac{450 - 238.23}{155.94}
$$

$$
z \approx 1.36
$$

This means that the customer spending 450 is approximately 1.36 standard deviations above the average customer spending.

Since the z-score is positive and relatively high, this customer spends significantly more than the average customer.

###  Business Insight

The customer spending 450 can be considered a potential high-value customer because their spending is substantially above the average of 238.23.

Such customers could be considered for **premium product recommendations, personalized offers, loyalty programs, or targeted cross-selling campaigns**.



> "The average customer spends $238.23, with a standard deviation of about $155.94. For a customer spending $450, the z-score is 1.36. This means the customer spends about 1.36 standard deviations above the average, indicating that they are a relatively high-spending customer and could be a potential high-value customer."

##  8. Probability of High Purchase Frequency or High Spending

###  Objective

Calculate the probability that a randomly selected customer:

- Makes **more than 5 purchases per month**, or
- Spends **more than $300 per month**.

We use the **addition rule of probability** because the customer can satisfy either condition or both conditions.

###  Addition Rule

For two events **A** and **B**:

$$
P(A \cup B) = P(A) + P(B) - P(A \cap B)
$$

Where:

- **A** = Customer makes more than 5 purchases
- **B** = Customer spends more than 300
- **A ∩ B** = Customer makes more than 5 purchases **and** spends more than 300

We subtract the intersection because customers satisfying **both conditions** would otherwise be counted twice.


```python
# Define the two conditions
A = df['no_of_purchases'] > 5
B = df['amount_spend'] > 300
```

```python
# Calculate individual probabilities
P_A = A.mean()
P_B = B.mean()
```

```python
# Calculate probability of both conditions
P_A_and_B = (A & B).mean()
```

```python
# Addition rule
P_A_or_B = P_A + P_B - P_A_and_B
```

```python
print("P(A) - More than 5 purchases:", P_A)
print("P(B) - Spending more than $300:", P_B)
print("P(A and B):", P_A_and_B)
print("P(A or B):", P_A_or_B)
print("Probability (%):", P_A_or_B * 100)
```

##  Interpretation

The probability that a randomly selected customer makes **more than 5 purchases per month** is **25.5%**, while the probability that a customer **spends more than 300 per month** is **28%**.

The probability that a customer satisfies **both conditions** is **13.9%**.

Using the addition rule:

$$
P(A \cup B) = P(A) + P(B) - P(A \cap B)
$$

Substituting the values:

$$
P(A \cup B) = 0.255 + 0.28 - 0.139
$$

$$
P(A \cup B) = 0.396
$$

Therefore, the probability that a randomly selected customer **either makes more than 5 purchases or spends more than 300** is:

$$
\boxed{39.6\%}
$$

###  Business Insight

There is a **39.6% probability** that a randomly selected customer belongs to this high-purchase or high-spending segment.

This indicates that a significant portion of customers demonstrate either **high purchase frequency or high spending**. These customers could be targeted with **personalized promotions, loyalty programs, product recommendations, and cross-selling campaigns**.


> "The probability of making more than 5 purchases is 25.5%, and the probability of spending more than $300 is 28%. Since some customers satisfy both conditions, we subtract their intersection of 13.9% using the addition rule. Therefore, the final probability is 39.6%. This means there is a 39.6% chance that a randomly selected customer either makes more than 5 purchases or spends more than 300."

##  9. 95% Confidence Interval for Customer Spending

###  Objective

Determine the **95% confidence interval** for the average amount spent by customers.

A confidence interval gives us a range of values within which the **true average customer spending** is likely to fall.

###  Formula

For a 95% confidence interval:

$$
CI = \text{Mean} \pm \text{Margin of Error}
$$

The margin of error is calculated as:

$$
\text{Margin of Error} = z_{critical} \times \frac{\text{Standard Deviation}}{\sqrt{n}}
$$

Where:

- **Mean** = Average customer spending
- **Standard Deviation** = Spread of customer spending
- **n** = Number of customers
- **z-critical = 1.96** for a 95% confidence level

Therefore:

$$
CI = \left(\text{Mean} - ME,\ \text{Mean} + ME\right)
$$

###  Business Objective

The confidence interval helps estimate the range in which the **true average customer spending** is likely to lie, based on the available customer sample.

```python
import math

# Calculate mean
mean_spend = df['amount_spend'].mean()

# Calculate sample standard deviation
std_spend = df['amount_spend'].std()

# Number of customers
n = df['amount_spend'].count()

# Z-critical value for 95% confidence
z_critical = 1.96

# Calculate standard error
standard_error = std_spend / math.sqrt(n)

# Calculate margin of error
margin_of_error = z_critical * standard_error

# Calculate confidence interval
lower_limit = mean_spend - margin_of_error
upper_limit = mean_spend + margin_of_error

print("Mean spending:", mean_spend)
print("Standard deviation:", std_spend)
print("Number of customers:", n)
print("Z-critical:", z_critical)
print("Margin of error:", margin_of_error)
print("95% Confidence Interval:", (lower_limit, upper_limit))
```

## 📊 Interpretation

The sample contains **1,000 customers** with an average monthly spending of **238.23**.

The standard deviation of customer spending is **155.94**, and for a 95% confidence level, the **z-critical value is 1.96**.

The calculated margin of error is approximately **9.67**.

Therefore, the 95% confidence interval is:

$$
CI = 238.23 \pm 9.67
$$

$$
CI = (228.56,\ 247.90)
$$

###  Final Result

The **95% confidence interval for the true average customer spending is 228.56 to 247.90**.

In simple terms, based on this sample, we estimate that the **true average monthly spending of customers is likely to fall between 228.56 and 247.90** at a 95% confidence level.

###  Business Insight

The estimated average customer spending is **238.23**, with a relatively narrow confidence interval of approximately **19.33 in total width**.

This gives the business a reasonable statistical range for estimating the average spending behavior of the overall customer population.

The result can be useful for **revenue planning, customer segmentation, marketing analysis, and setting spending-based customer targets**.


> "We have 1,000 customers, and their average spending is 238.23. Using a standard deviation of 155.94 and a z-critical value of 1.96, the margin of error is about 9.67. Therefore, the 95% confidence interval is 228.56 to 247.90. This means we estimate that the true average customer spending lies within this range at a 95% confidence level."

##  10. Probability of High Cross-Sell Conversion Among Frequent Customers
Determine the probability that a customer who makes **5 or more purchases per month** has a **high cross-sell conversion rate greater than 80%**.

This is a **conditional probability** problem because we are considering only customers who make 5 or more purchases.

###  Define the Events

Let:

- **A** = Customer makes **5 or more purchases**
- **B** = Customer has a cross-sell conversion rate **greater than 80%**

We need to calculate:

$$
P(B|A)
$$

This means:

> Probability of having a high cross-sell conversion rate **given that the customer makes 5 or more purchases**.

###  Conditional Probability Formula

$$
P(B|A) = \frac{P(A \cap B)}{P(A)}
$$

In terms of customer counts:

$$
P(B|A) =
\frac{\text{Number of customers satisfying both conditions}}
{\text{Number of customers making 5 or more purchases}}
$$

###  Business Objective

This analysis helps determine whether **frequent purchasers are also strong cross-sell customers**.

If the probability is high, customers with frequent purchases may represent an important segment for **cross-selling and personalized product recommendations**.

```python
# Step 1: P(A)
P_A = (df['cross_sell_conversion_rate'] > 0.80).mean()

# Step 2: P(B)
P_B = (df['no_of_purchases'] >= 5).mean()

# Step 3: P(B|A)
P_B_given_A = (
    df.loc[df['cross_sell_conversion_rate'] > 0.80, 'no_of_purchases'] >= 5
).mean()

# Step 4: Bayes' Theorem
probability = (P_B_given_A * P_A) / P_B

print(f"{probability:.3f}")
```

## 📊 Interpretation

The probability that a customer who makes **5 or more purchases per month** has a **cross-sell conversion rate greater than 80%** is:

$$
P(B|A) = 0.3506
$$

or approximately:

$$
\boxed{35.06\%}
$$

This means that among customers who make **5 or more purchases per month, approximately 35.06% have a high cross-sell conversion rate (greater than 80%)**.

###  Business Insight

The result indicates that a noticeable proportion of frequent purchasers also have a high likelihood of purchasing additional products.

These customers can be considered a potentially valuable segment for **cross-selling campaigns**, such as:

- Personalized product recommendations
- Complementary product offers
- Bundle offers
- Loyalty rewards
- Targeted promotional campaigns

However, since approximately **64.94%** of frequent purchasers do not have a cross-sell conversion rate above 80%, purchase frequency alone may not be sufficient to identify high cross-sell customers. Other customer characteristics could also be considered for better targeting.



> "We calculated the conditional probability because we are interested only in customers who make 5 or more purchases. Among those customers, the probability of having a cross-sell conversion rate above 80% is 35.06%. So, approximately 35% of frequent purchasers are also high cross-sell customers."
